```python
# ==========================================================
# Core Data Handling & Numerical Computing
# ==========================================================
import os
import warnings
import numpy as np
import pandas as pd

# ==========================================================
# Data Visualization
# ==========================================================
import matplotlib.pyplot as plt
import seaborn as sns

# ==========================================================
# Preprocessing & Dataset Splitting
# ==========================================================
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder

# ==========================================================
# Machine Learning Models
# ==========================================================
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.discriminant_analysis import (
    LinearDiscriminantAnalysis,
    QuadraticDiscriminantAnalysis,
)
from xgboost import XGBClassifier

# ==========================================================
# Model Evaluation & Metrics
# ==========================================================
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
)

# ==========================================================
# Visualization Settings
# ==========================================================
%matplotlib inline
sns.set_theme(style="whitegrid")
warnings.filterwarnings("ignore")
```

```python
data1= pd.read_csv('outputs/baseline_features.csv')
data2= pd.read_csv('outputs/extended_features.csv')
data2.head()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
```

```python
train_data1, test_data1 = train_test_split(data1, test_size=0.2, random_state=42)
train_data2, test_data2 = train_test_split(data2, test_size=0.2, random_state=42)
```

```python
metadata_rows=['filename','classID','fold','original_sampling_rate','processed_sampling_rate','duration']

def preprocess_data(data):
    data = data.drop(metadata_rows, axis=1)
    return data
```

```python
data1['duration'].describe()
```

```python
train_data1_updated = preprocess_data(train_data1)
test_data1_updated = preprocess_data(test_data1)

train_data2_updated = preprocess_data(train_data2)
test_data2_updated = preprocess_data(test_data2)
```

```python
train_data1_updated.head()
```

```python
train_data1_updated.reset_index(drop=True, inplace=True)
train_data1_updated.head()
```

```python
test_data1_updated.reset_index(drop=True, inplace=True)
train_data2_updated.reset_index(drop=True, inplace=True)
test_data2_updated.reset_index(drop=True, inplace=True)
```

```python
encoder = LabelEncoder()
train_data1_updated['target']= encoder.fit_transform(train_data1_updated['class'])
test_data1_updated['target']= encoder.transform(test_data1_updated['class'])
train_data2_updated['target']= encoder.transform(train_data2_updated['class'])
test_data2_updated['target']= encoder.transform(test_data2_updated['class'])
```

```python
scaler = StandardScaler()
train_data1_scaled = scaler.fit_transform(train_data1_updated.drop(['class','target'], axis=1))
test_data1_scaled = scaler.transform(test_data1_updated.drop(['class','target'], axis=1))

train_data2_scaled = scaler.fit_transform(train_data2_updated.drop(['class','target'], axis=1))
test_data2_scaled = scaler.transform(test_data2_updated.drop(['class','target'], axis=1))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
```

```python
model1= LogisticRegression(max_iter=1000)
model2= SVC()
model3= RandomForestClassifier(n_estimators=100)
```

```python
X_train1, y_train1 = train_data1_scaled, train_data1_updated['target']
X_test1, y_test1 = test_data1_scaled, test_data1_updated['target']
X_train2, y_train2 = train_data2_scaled, train_data2_updated['target']
X_test2, y_test2 = test_data2_scaled, test_data2_updated['target']
```

```python
model1.fit(X_train1, y_train1)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
)
```

```python
y_pred1 = model1.predict(X_test1)

accuracy1 = accuracy_score(y_test1, y_pred1)
precision1 = precision_score(y_test1, y_pred1, average='macro')
recall1 = recall_score(y_test1, y_pred1, average='macro')
f11 = f1_score(y_test1, y_pred1, average='macro')

print("Model 1 Results:")
print(f"Accuracy: {accuracy1:.2f}")
print(f"Precision: {precision1:.2f}")
print(f"Recall: {recall1:.2f}")
print(f"F1-Score: {f11:.2f}")
```

```python
y_pred1_svm = model2.predict(X_test1)

accuracy1_svm = accuracy_score(y_test1, y_pred1_svm)
precision1_svm = precision_score(y_test1, y_pred1_svm, average='macro')
recall1_svm = recall_score(y_test1, y_pred1_svm, average='macro')
f11_svm = f1_score(y_test1, y_pred1_svm, average='macro')

print("Model 2 (SVM) Results:")
print(f"Accuracy: {accuracy1_svm:.2f}")
print(f"Precision: {precision1_svm:.2f}")
print(f"Recall: {recall1_svm:.2f}")
print(f"F1-Score: {f11_svm:.2f}")
```

```python
model3.fit(X_train1, y_train1)
```

```python
y_pred1_rf = model3.predict(X_test1)
```

```python
accuracy1_rf = accuracy_score(y_test1, y_pred1_rf)
precision1_rf = precision_score(y_test1, y_pred1_rf, average='macro')
recall1_rf = recall_score(y_test1, y_pred1_rf, average='macro')
f11_rf = f1_score(y_test1, y_pred1_rf, average='macro')

print("Model 3 (Random Forest) Results:")
print(f"Accuracy: {accuracy1_rf:.2f}")
print(f"Precision: {precision1_rf:.2f}")
print(f"Recall: {recall1_rf:.2f}")
print(f"F1-Score: {f11_rf:.2f}")
```

```python
from xgboost import XGBClassifier

model4 = XGBClassifier(eval_metric='mlogloss', random_state=42)
model4.fit(X_train1, y_train1)

y_pred1_xgb = model4.predict(X_test1)

accuracy1_xgb = accuracy_score(y_test1, y_pred1_xgb)
precision1_xgb = precision_score(y_test1, y_pred1_xgb, average='macro')
recall1_xgb = recall_score(y_test1, y_pred1_xgb, average='macro')
f11_xgb = f1_score(y_test1, y_pred1_xgb, average='macro')

print("Model 4 (XGBoost) Results:")
print(f"Accuracy: {accuracy1_xgb:.2f}")
print(f"Precision: {precision1_xgb:.2f}")
print(f"Recall: {recall1_xgb:.2f}")
print(f"F1-Score: {f11_xgb:.2f}")
```

```python
# ==========================================================
# Model 5: Linear Discriminant Analysis (LDA)
# ==========================================================
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

model5_lda = LinearDiscriminantAnalysis()
model5_lda.fit(X_train1, y_train1)

y_pred1_lda = model5_lda.predict(X_test1)

accuracy1_lda = accuracy_score(y_test1, y_pred1_lda)
precision1_lda = precision_score(y_test1, y_pred1_lda, average='macro')
recall1_lda = recall_score(y_test1, y_pred1_lda, average='macro')
f11_lda = f1_score(y_test1, y_pred1_lda, average='macro')

print("Model 5 (LDA) Results:")
print(f"Accuracy: {accuracy1_lda:.2f}")
print(f"Precision: {precision1_lda:.2f}")
print(f"Recall: {recall1_lda:.2f}")
print(f"F1-Score: {f11_lda:.2f}")

print("\nLDA Explained Variance Ratio across Discriminants:")
for i, var in enumerate(model5_lda.explained_variance_ratio_):
    print(f"LD{i+1}: {var:.4f}")
```

```python
# ==========================================================
# Model 6: Quadratic Discriminant Analysis (QDA)
# ==========================================================
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis

model6_qda = QuadraticDiscriminantAnalysis()
model6_qda.fit(X_train1, y_train1)

y_pred1_qda = model6_qda.predict(X_test1)

accuracy1_qda = accuracy_score(y_test1, y_pred1_qda)
precision1_qda = precision_score(y_test1, y_pred1_qda, average='macro')
recall1_qda = recall_score(y_test1, y_pred1_qda, average='macro')
f11_qda = f1_score(y_test1, y_pred1_qda, average='macro')

print("Model 6 (QDA) Results:")
print(f"Accuracy: {accuracy1_qda:.2f}")
print(f"Precision: {precision1_qda:.2f}")
print(f"Recall: {recall1_qda:.2f}")
print(f"F1-Score: {f11_qda:.2f}")
```

```python
# ==========================================================
# Models Performance Comparison Summary
# ==========================================================
models_summary = pd.DataFrame({
    "Model": ["Logistic Regression", "SVM", "Random Forest", "XGBoost", "LDA", "QDA"],
    "Accuracy": [accuracy1, accuracy1_svm, accuracy1_rf, accuracy1_xgb, accuracy1_lda, accuracy1_qda],
    "Precision": [precision1, precision1_svm, precision1_rf, precision1_xgb, precision1_lda, precision1_qda],
    "Recall": [recall1, recall1_svm, recall1_rf, recall1_xgb, recall1_lda, recall1_qda],
    "F1-Score": [f11, f11_svm, f11_rf, f11_xgb, f11_lda, f11_qda]
})

print("=== Summary Comparison on Baseline Features ===")
print(models_summary.sort_values(by="F1-Score", ascending=False).to_string(index=False))
```
